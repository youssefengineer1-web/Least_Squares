```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(-5, 5, 1000)
y = np.linspace(-5, 5, 1000)

X, Y = np.meshgrid(x, y)
Z = np.zeros((1000, 1000))

fig = plt.figure(figsize=(12, 12))
ax = fig.add_subplot(111, projection='3d', computed_zorder=False)

ax.plot_surface(X, Y, Z, color="lightblue")

ax.quiver(0, 0, 0, 3, 3, 3, arrow_length_ratio=0.1, linewidth=2,color='red')
ax.quiver(3, 3, 0, 0, 0, 3, arrow_length_ratio=0.1, linewidth=2, )
ax.quiver(0, 0, 0, 3, 3, 0, arrow_length_ratio=0.1, linewidth=2, zorder=5, color='green')

ax.text(2, 1.5, 2.5, r'$\vec{b}$', fontsize=18)
ax.text(3, 3.5, 2.5, r"$\vec{b}$-$𝑨\vec{𝑥}$*", fontsize=18)
ax.text(2, 1, 0, r"$𝑨\vec{𝑥}$*", fontsize=18)
ax.text(-1, -1.5, 0, r"$Col(𝑨)$", fontsize=18)

ax.set_zlim(0, 4)
ax.set_zticks(np.arange(5))

plt.show()
```
